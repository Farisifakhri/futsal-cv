```python
import torch
import numpy as np

# 1. FIX: Bypass PyTorch 2.6+ UnpicklingError
_original_torch_load = torch.load
def _patched_torch_load(*args, **kwargs):
    kwargs.setdefault('weights_only', False)
    return _original_torch_load(*args, **kwargs)
torch.load = _patched_torch_load

# 2. FIX: Kompatibilitas numpy 2.0+ trapz
if not hasattr(np, 'trapz'):
    np.trapz = np.trapezoid

from ultralytics import YOLO

# 3. Load model terbaik hasil training
model = YOLO("models/best_futsal_yolov8m_v3.pt")

# 4. Jalankan Evaluasi Murni pada data TEST
metrics = model.val(data="data_augmented/data.yaml", split="test")

# 5. Tampilkan Hasil mAP Data Test
print("\n" + "=" * 50)
print(f"📊 Overall mAP50 (Data Test)     : {metrics.box.map50:.3f}")
print(f"📊 Overall mAP50-95 (Data Test)  : {metrics.box.map:.3f}")
print("=" * 50)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# ============================================================
# 1. Path ke file results.csv masing-masing model
# ============================================================
# Ganti path ini sesuai folder penyimpanan hasil training Anda
csv_non_aug = "futsal/yolov8m_futsal_v1/results.csv"  # Hasil v1 (Tanpa Augmentasi)
csv_aug     = "futsal/yolov8m_futsal_v3/results.csv"  # Hasil v3 (Dengan Augmentasi)

# Read CSV files
df_non = pd.read_csv(csv_non_aug)
df_aug = pd.read_csv(csv_aug)

# Clean column names (menghapus spasi di nama kolom)
df_non.columns = [c.strip() for c in df_non.columns]
df_aug.columns = [c.strip() for c in df_aug.columns]

# ============================================================
# 2. Konfigurasi Subplots (2 baris x 3 kolom)
# ============================================================
fig, axes = plt.subplots(3, 2, figsize=(14, 15))
fig.suptitle('Gambar 4. Grafik Perbandingan Performa YOLOv8m (Non-Augmentasi vs Augmentasi)', 
             fontsize=14, fontweight='bold', y=0.99)

# Grid style
for ax_row in axes:
    for ax in ax_row:
        ax.grid(True, linestyle='--', alpha=0.6)

# ------------------------------------------------------------
# 1. Train Box Loss
# ------------------------------------------------------------
ax = axes[0, 0]
ax.plot(df_non['epoch'], df_non['train/box_loss'], color='#4169E1', label='Non Augmentasi - Train')
ax.plot(df_non['epoch'], df_non['val/box_loss'], color='#32CD32', linestyle='--', label='Non Augmentasi - Val')
ax.plot(df_aug['epoch'], df_aug['train/box_loss'], color='#DC143C', label='Augmentasi - Train')
ax.plot(df_aug['epoch'], df_aug['val/box_loss'], color='#BA55D3', linestyle='--', label='Augmentasi - Val')
ax.set_title('Train Box Loss (mean ± std)', fontsize=11)
ax.set_xlabel('Epoch')
ax.set_ylabel('train/box_loss')
ax.legend(loc='upper right', fontsize=8)

# ------------------------------------------------------------
# 2. Train Cls Loss
# ------------------------------------------------------------
ax = axes[0, 1]
ax.plot(df_non['epoch'], df_non['train/cls_loss'], color='#4169E1', label='Non Augmentasi - Train')
ax.plot(df_non['epoch'], df_non['val/cls_loss'], color='#32CD32', linestyle='--', label='Non Augmentasi - Val')
ax.plot(df_aug['epoch'], df_aug['train/cls_loss'], color='#DC143C', label='Augmentasi - Train')
ax.plot(df_aug['epoch'], df_aug['val/cls_loss'], color='#BA55D3', linestyle='--', label='Augmentasi - Val')
ax.set_title('Train Cls Loss (mean ± std)', fontsize=11)
ax.set_xlabel('Epoch')
ax.set_ylabel('train/cls_loss')
ax.legend(loc='upper right', fontsize=8)

# ------------------------------------------------------------
# 3. Precision (B)
# ------------------------------------------------------------
ax = axes[1, 0]
ax.plot(df_non['epoch'], df_non['metrics/precision(B)'], color='#4169E1', label='Non Augmentasi')
ax.plot(df_aug['epoch'], df_aug['metrics/precision(B)'], color='#32CD32', label='Augmentasi')
ax.set_title('Metrics Precision(B) (mean ± std)', fontsize=11)
ax.set_xlabel('Epoch')
ax.set_ylabel('metrics/precision(B)')
ax.legend(loc='lower right', fontsize=9)

# ------------------------------------------------------------
# 4. Recall (B)
# ------------------------------------------------------------
ax = axes[1, 1]
ax.plot(df_non['epoch'], df_non['metrics/recall(B)'], color='#4169E1', label='Non Augmentasi')
ax.plot(df_aug['epoch'], df_aug['metrics/recall(B)'], color='#32CD32', label='Augmentasi')
ax.set_title('Metrics Recall(B) (mean ± std)', fontsize=11)
ax.set_xlabel('Epoch')
ax.set_ylabel('metrics/recall(B)')
ax.legend(loc='lower right', fontsize=9)

# ------------------------------------------------------------
# 5. mAP50 (B)
# ------------------------------------------------------------
ax = axes[2, 0]
ax.plot(df_non['epoch'], df_non['metrics/mAP50(B)'], color='#4169E1', label='Non Augmentasi')
ax.plot(df_aug['epoch'], df_aug['metrics/mAP50(B)'], color='#32CD32', label='Augmentasi')
ax.set_title('Metrics Map50(B) (mean ± std)', fontsize=11)
ax.set_xlabel('Epoch')
ax.set_ylabel('metrics/mAP50(B)')
ax.legend(loc='lower right', fontsize=9)

# ------------------------------------------------------------
# 6. mAP50-95 (B)
# ------------------------------------------------------------
ax = axes[2, 1]
ax.plot(df_non['epoch'], df_non['metrics/mAP50-95(B)'], color='#4169E1', label='Non Augmentasi')
ax.plot(df_aug['epoch'], df_aug['metrics/mAP50-95(B)'], color='#32CD32', label='Augmentasi')
ax.set_title('Metrics Map50-95(B) (mean ± std)', fontsize=11)
ax.set_xlabel('Epoch')
ax.set_ylabel('metrics/mAP50-95(B)')
ax.legend(loc='lower right', fontsize=9)

plt.tight_layout()
plt.savefig('grafik_perbandingan_yolov8m.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Grafik berhasil disimpan ke 'grafik_perbandingan_yolov8m.png'")
```
